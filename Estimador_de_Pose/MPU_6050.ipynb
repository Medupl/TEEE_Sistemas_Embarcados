#### Acelerômetro e Giroscópio 3 Eixos MPU-6050

O MPU 6050 é um sensor de acelerômetro e giroscópio de seis eixos integrados em um único chip, oferecendo 6 graus de liberdade (6DOF). Com três eixos para o acelerômetro e três eixos para o giroscópio, proporciona uma ampla gama de detecção de movimento e orientação. Além disso, o módulo GY-521 inclui um sensor de temperatura embutido, permitindo medições precisas na faixa de -40°C a +85°C. 

Sua alta precisão é garantida pelo conversor analógico-digital de 16 bits para cada canal, capturando simultaneamente os canais X, Y e Z. O MPU-6050 é amplamente utilizado em uma variedade de aplicações, incluindo dispositivos de navegação inercial, estabilização de câmeras, robótica e controle de movimento em jogos e realidade virtual.

#### Como funciona o Acelerômetro e Giroscópio 3 Eixos MPU-6050?

O MPU 6050, um sensor de seis eixos, compreende um acelerômetro e um giroscópio MEMS (Micro-Electro-Mechanical Systems) integrados em um único chip. Funciona medindo as forças de aceleração lineares em três eixos (X, Y e Z) com o acelerômetro e as taxas de rotação em torno desses mesmos eixos com o giroscópio.

O acelerômetro detecta mudanças na velocidade de movimento, enquanto o giroscópio mede a velocidade angular ou taxa de rotação. Essas informações combinadas fornecem dados precisos sobre a orientação e movimento do dispositivo em todas as direções tridimensionais. Esses dados são então processados e convertidos em sinais digitais pelo MPU-6050, tornando-os disponíveis para serem lidos e utilizados por um microcontrolador ou outro dispositivo de processamento de dados.

#### Pinagem do Acelerômetro e Giroscópio 3 Eixos MPU-6050

O sensor MPU6050 é comumente encontrado em um módulo como o GY-521, que possui pinos para facilitar a conexão. Os pinos típicos encontrados no MPU-6050 são:

- VCC: Pino de alimentação, geralmente conectado a uma fonte de energia de 3.3V ou 5V.
- GND: Pino de terra, conectado ao terra do sistema.
- SCL: Pino de Clock Serial, usado para comunicação I2C.
- SDA: Pino de Dados Serial, também usado para comunicação I2C.
- XDA: Saída de dados auxiliar para comunicação I2C.
- XCL: Clock auxiliar para comunicação I2C.
- AD0: Pino de endereço. Pode ser usado para alterar o endereço I2C do dispositivo.
- INT: Pino de interrupção, usado para notificar o microcontrolador sobre eventos como detecção de movimento.
